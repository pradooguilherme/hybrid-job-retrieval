```python
import pandas as pd

df = pd.read_csv('../dataset/job.csv')
```

```python
display(df.head(5))
```

```python
jobs_to_look = ['HR','DESIGNER','INFORMATION-TECHNOLOGY',
                'TEACHER','ADVOCATE','BUSINESS-DEVELOPMENT',
                'HEALTHCARE','FITNESS','AGRICULTURE',
                'BPO','SALES','CONSULTANT','DIGITAL-MEDIA',
                'AUTOMOBILE','CHEF','FINANCE',
                'APPAREL','ENGINEERING','ACCOUNTANT',
                'CONSTRUCTION','PUBLIC-RELATIONS', 'BANKING',
                'ARTS','AVIATION']
```

```python
for job in jobs_to_look:
    num_same = df[df['position_title'].str.lower().str.contains(job.lower(), na=False)].shape[0]
    if num_same < 3:
        jobs_to_look.remove(job)

print(f"Jobs (from resume table) with 3 or more matches in position_title (from job table): {str(jobs_to_look)}")
```

```python
# Removing unnecessary columns
df.drop(columns=['company_name','model_response'], inplace=True)
```

```python
# Checking null strings
df.isnull().sum()
```

```python
df['category'] = 'no_match'

for job in jobs_to_look:
    filter = df['position_title'].str.contains(job, case=False, na=False)
    df.loc[filter, 'category'] = job

filter = df[df['category'] == 'no_match']
df = df.drop(filter.index)

print(f"Job table after removing no match jobs: {df.shape[0]}")
```

```python
display(df.head(15))
```

```python
import re
import string

df['content'] = df['position_title'] + ' ' + df['job_description']

def preprocess(txt):
    txt = str(txt).lower()
    txt = re.sub(r"http\S+", " ", txt)
    txt = re.sub(r"@\w+", " ", txt)
    txt = re.sub(r"[^a-zA-Z ]", " ", txt)
    txt = txt.translate(str.maketrans('', '', string.punctuation))
    txt = " ".join(txt.split())

    return txt

df['content'] = df['content'].apply(preprocess)
```

```python
save_path = '../dataset/cleaned/job.csv'
df.to_csv(save_path, index=False)

print(f"Saving job table to {save_path}")
```
