```python
import pandas as pd
import re
import string

df = pd.read_csv('../dataset/Resume.csv', index_col='ID')

print(df.info())
```

```python
# Removing unnecessary column
df = df.drop(columns=['Resume_html'])

# Removing empty lines
df = df[df['Resume_str'].str.strip().str.len() > 0]
```

```python
df['Category'].unique()
```

```python
keeping_columns = ['HR', 'DESIGNER', 'TEACHER', 'ADVOCATE', 'HEALTHCARE', 'AGRICULTURE', 'SALES', 'CONSULTANT', 'AUTOMOBILE', 'FINANCE', 'ENGINEERING', 'ACCOUNTANT', 'CONSTRUCTION', 'BANKING', 'AVIATION']

df =  df[df['Category'].isin(keeping_columns)]
```

```python
def preprocess(txt):
    txt = str(txt).lower()
    txt = re.sub(r"http\S+", " ", txt)
    txt = re.sub(r"@\w+", " ", txt)
    txt = re.sub(r"[^a-zA-Z ]", " ", txt)
    txt = txt.translate(str.maketrans('', '', string.punctuation))
    txt = " ".join(txt.split())

    return txt
```

```python
df['resume'] = df['Resume_str'].apply(preprocess)
```

```python
print(df.head(3))
df.to_csv('dataset/cleaned/resume.csv', index=False)
```
